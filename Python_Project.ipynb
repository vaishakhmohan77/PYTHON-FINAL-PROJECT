**2. DATA PREPROCESSING**

**A) HANDLING MISSING VALUES :-** Missing values are identified in the dataset to determine which columns require further treatment during the data cleaning process.



```python
df.isnull().sum()

# PURPOSE - isnull() checks whether each value is missing and sum() counts the missing values in each column.
```

Result :- (Missing Value Analysis)

The missing value analysis shows that 5 columns contain missing values: First response time (18), Resolution time (418), Close time (1,157), Agent interactions (18), and Survey results (1,157). The remaining columns have no missing values.


**A.(1) Handling Missing Values: First Response Time (Column)**
                               
                               
The 'First response time' column contains 18 missing values. The column is converted to datetime format, and the median timestamp is used to replace the missing values.

```python
df['First response time'] = df['First response time'].fillna(df['First response time'].mode()[0])
print("Missing values:", df['First response time'].isnull().sum())
```

Result :- The 18 missing values in the `First response time` column were successfully replaced using the mode. The column now contains no missing values.

**A.(2) Handling Missing Values: Resolution time (Column)**

The 'Resolution time' column contains 418 missing values. Since the column is stored as an object data type, the mode is used to replace the missing values.

```python
df['Resolution time'] = df['Resolution time'].fillna(df['Resolution time'].mode()[0])
print("Missing values:", df['Resolution time'].isnull().sum())
```

Result :- The 418 missing values in the 'Resolution time' column were successfully replaced using the mode. The column now contains no missing values.

**A.(3) Handling Missing Values:- Close time (Column)**

The 'Close time' column contains 1,157 missing values. Since the column is stored as an object data type, the mode is used to replace the missing values.

```python
df['Close time'] = df['Close time'].fillna(df['Close time'].mode()[0])
print("Missing values:", df['Close time'].isnull().sum())
```

Result :- The 1,157 missing values in the `Close time` column were successfully replaced using the mode. The column now contains no missing values.

**A.(4) Handling Missing Values: Agent Interactions (Column)**

The 'Agent interactions' column contains 18 missing values. Since this is a numerical column, the median is used to replace the missing values.

```python
df['Agent interactions'] = df['Agent interactions'].fillna(df['Agent interactions'].median())
print("Missing values:", df['Agent interactions'].isnull().sum())
```

Result :- The 18 missing values in the 'Agent interactions' column were successfully replaced using the median. The column now contains no missing values.

**A.(5) Handling Missing Values: Survey Results (Column)**

The 'Survey results' column contains 1,157 missing values. Since this is a numerical column, the median is used to replace the missing values.

```python
df['Survey results'] = df['Survey results'].fillna(df['Survey results'].median())
print("Missing values:", df['Survey results'].isnull().sum())
```

Result :- The 1,157 missing values in the 'Survey results' column were successfully replaced using the median. The column now contains no missing values.

**FINAL MISSING VALUE CHECK**

A final check was performed to confirm that all missing values in the dataset were handled successfully.

**C) Correcting data types :-**

The data types of the columns were checked and corrected where necessary. The date columns were converted to datetime format because they contain dates and times in day-month-year format.

The 'First response time', 'Resolution time', and 'Close time' columns contain time durations in minutes and seconds. Therefore, these columns were converted to timedelta format.

Correcting the data types makes the dataset suitable for date, time, and duration analysis.

```python
df['Created time'] = pd.to_datetime(df['Created time'], dayfirst=True)
df['Expected SLA to resolve'] = pd.to_datetime(df['Expected SLA to resolve'], dayfirst=True)
df['Expected SLA to first response'] = pd.to_datetime(df['Expected SLA to first response'], dayfirst=True)

df['First response time'] = pd.to_timedelta('00:' + df['First response time'])
df['Resolution time'] = pd.to_timedelta('00:' + df['Resolution time'])
df['Close time'] = pd.to_timedelta('00:' + df['Close time'])

print(df.dtypes)
```

**Result :-** The data types were successfully corrected. Date and time columns are now stored in datetime format, while duration columns are stored in timedelta format. The numerical columns remain in numerical format.

The dataset is now ready for further data analysis.
